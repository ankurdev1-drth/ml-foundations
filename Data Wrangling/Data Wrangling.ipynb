```python
df2
```

```python
#example of many to one join because df1 has multiple rows labeled a and b whereas df2 has only one row for each value in key column 
pd.merge(df1,df2)
```

```python
#For merging/joining operations and combine datasets by linking rows using one/more keys we use  "pandas.merge" function
```

```python
#Int 64 type for nullable integers
#also here its not specified which column to join on due to which pandas.merge used the overlapping names as keys by default 
```

```python
pd.merge(df1, df2, on = "key") #have confusion here !
```

```python
#NOTE FOR confusion of on = "key"   :The on="key" parameter tells pandas which column to use as the joining condition — it must exist in both DataFrames
```

```python
#the order of column output is unspecified in pandas.merge function 
```

```python
df3 = pd.DataFrame({"lkey": ["b","b", "a", "c", "a", "a", "b"], 
                    "data1":pd.Series(range(7), dtype="Int64")})
```

```python
df4 = pd.DataFrame({"rkey": ["a","b","d"],
                    "data2": pd.Series(range(3), dtype="Int64")})
```

```python
pd.merge(df3,df4, left_on = "lkey", right_on = "rkey")
```

```python
#here pd.merge(df3,df4) will give syntax error as there would be no common values to perform merge
#also "C" and "d" are missing because the result that we got above is the common set found in both the tables 
#and also because pandas.merge does an "inner" join
```

```python
pd.merge(df1,df2, how="outer")
```

```python
#using how  we told pandas to to combine the effect of both left and right joins 
```

```python
pd.merge(df1, df2, how="left")
```

```python
pd.merge(df1,df2, how ="right")
```

**Pivoting “Long” to “Wide” Format

```python
#In long or stack format individual values are represented by a single row in a table rathen than multiple values per row
```

```python
L = pd.DataFrame({
    'student': ['Alice','Alice','Bob','Bob'],
    'subject': ['Math','English','Math','English'],
    'score':   [90, 85, 72, 88]
})
```

```python
L
```

```python
wide = L.pivot(
    index='student',    # stays as rows
    columns='subject',  # values → column headers
    values='score'      # fills the cells
)
```

```python
wide
```

```python
wide.reset_index(inplace=True)
```

```python
wide
```

```python
wide.columns.name = None #cleaning up the names
```

```python
wide
```

```python
#aggregate duplicates
```

```python
wide = L.pivot_table(
    index='student',
    columns='subject',
    values='score',
    aggfunc='mean'   # 'sum', 'max', 'count', etc.
).reset_index()
```

```python
wide
```

```python
Lf = pd.DataFrame({"key": ["foo", "bar", "baz"],
                           "A": [1, 2, 3],
                   "B": [4,5,6],
                   "C": [7,8,9]})
```

```python
Lf
```

```python
#When using pandas.melt, we must indicate which columns (if any) are group indicators
```

```python
melt = pd.melt(Lf, id_vars = "key")  
#here note that id_vars is a argument (parameter) of the function pd.melt() 
#id_vars

# stands for:

# identifier variables

# Meaning:

# columns that should stay as identifiers and should NOT be melted
```

```python
melt
```

```python
#reshaping it back to the original one
reshaped = melt.pivot(index="key",columns="variable",values="value")
```

```python
reshaped
```

```python
#using reset_index to move the data back into a column
```

```python
reshaped.reset_index()
```

```python
#we can also specify a subset of columns to use as value columns
```

```python
pd.melt(Lf, id_vars="key", value_vars=["A", "B"])
```

```python
#pd.melts can be used without any identifiers 
```

```python
 pd.melt(Lf, value_vars=["A", "B", "C"])
```
